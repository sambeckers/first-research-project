```python
"""
photoz_gds
Created on 18-02-2024

@author(s): Sam Beckers

Eazy photo-z fitting for the GDS catalog
"""
import os
os.chdir('/Users/sam/eazy-photoz') # Change to the eazy-photoz directory
print(os.getcwd())
import numpy as np
import matplotlib.pyplot as plt
plt.rcParams.update({
    "text.usetex": True,
    "font.family": "Times New Roman",
    "font.sans-serif": "helvetica"
})
import eazy

# Suppress warnings
import warnings
from astropy.utils.exceptions import AstropyWarning
np.seterr(all='ignore')
warnings.simplefilter('ignore', category=AstropyWarning)

f_path = '/Users/sam/Fresco/' # Path to the FRESCO directory

# Define the parameters for the EAZY fitting
params = {}
params['CATALOG_FILE'] = f_path + 'Catalogs_v2/gds_catalog_filtered.cat'
params['CATALOG_FORMAT'] = 'ascii' # important to specify the format
params['OUTPUT_DIRECTORY'] = f_path + 'eazy outputs'
params['MAIN_OUTPUT_FILE'] = f_path + 'eazy outputs/gds_photoz.eazypy'
params['N_MIN_COLORS'] = 3 # Minimum number of colors
params['NOT_OBS_THRESHOLD'] = -90 # Threshold for non-detection

params['Z_MAX'] = 15 # Maximum redshift
params['Z_STEP'] = 0.005 # Redshift step
params['PRIOR_ABZP'] = 23.9 # AB zeropoint
params['PRIOR_FILTER'] = 375
params['MW_EBV'] = 0.1909 # Milky Way E(B-V) reddening
params['CAT_HAS_EXTCORR'] = False # Catalog has extinction correction

# Planck flat lambda CDM cosmology (Plank Colloboration et al. 2020)
params['H0'] = 67.36
params['OMEGA_M'] = 0.3153
params['OMEGA_L'] = 0.6847 

params['WAVELENGTH_FILE'] = 'templates/uvista_nmf/lambda.def'
params['PRIOR_FILE'] = 'templates/prior_F160W_TAO.dat'
params['TEMPLATES_FILE'] = 'templates/sfhz/corr_sfhz_13.param'
params['TEMP_ERR_FILE'] = 'templates/template_error_cosmos2020.txt'
params['TEMP_ERR_A2'] = 1. # Template error amplitude
params['SYS_ERR'] = 0.05 # Systematic error

params['FILTERS_RES'] = 'filters/FILTER.RES.latest'
translate_file = 'inputs/zphot.translate'

# Initialize the photo-z object
self = eazy.photoz.PhotoZ(param_file=None, translate_file=translate_file, zeropoint_file=None, 
                          params=params, load_prior=True, load_products=False)

# Fit the catalog to the templates
self.fit_catalog()

# Save the results
self.standard_output()
```

```python
from matplotlib.lines import Line2D
from astropy.io import fits
from astropy import units as u
from astropy.wcs import WCS
from astropy.coordinates import SkyCoord
from astropy.nddata import Cutout2D
from matplotlib.gridspec import GridSpec
import numpy as np
import matplotlib.pyplot as plt
plt.rcParams.update({
    "text.usetex": True,
    "font.family": "Times New Roman",
    "font.sans-serif": "helvetica"
})
from tqdm import tqdm
from pathlib import Path

# Show the fit
fig_path = Path('/Users/sam/Documents/GitHub/FRP/Figures/')
f_path = Path('/Users/sam/FRESCO/')
cat = np.genfromtxt(f_path / 'Catalogs_v2/gds_zphot_catalog_filtered.cat', delimiter=' ', names=True, comments='#')
zout = fits.open(f_path / 'eazy outputs/gds_photoz.eazypy.zout.fits')
id = zout[1].data['id']
nusefilt = zout[1].data['nusefilt']
images_data = []
with open(f_path / 'gds-sci-filenames_ordered_incl_f444w.txt', 'r') as catalog_file:
    filenames = catalog_file.read().splitlines()
    for f in filenames:
        hdul = fits.open(f_path / f)
        images_data.append(hdul[0].data)
f_names = ['F336WU', 'F435W', 'F475W', 'F606W', 'F606WU', 'F775W', 'F814W', 'F814WU', 'F850LP', 'F850LPU', 'F105W', 'F110W', 'F125W', 'F140W', 'F160W', 'F182M', 'F210M', 'F430M', 'F460M', 'F480M', 'F444W']

match_template_arr = []
for ID, RA, DEC in tqdm(zip(cat['ID'], cat['ra'], cat['dec'])):
    # if ID == 5160.0:
    print(ID)
    with plt.ioff():
        fig, dir = self.show_fit(float(ID), template_color='red', logpz=False, add_label=False)
        
        ix = self.idx[self.OBJID == ID][0]
        z = self.zbest[ix]

        # Calculate the reduced chi^2
        zout_idx = np.where(id == ID)[0][0]
        nusefilt_id = nusefilt[zout_idx]
        reduced_chi2 = dir['chi2'] / nusefilt_id

        fig.set_dpi(450)
        fig.get_axes()[0].set_ylabel(r'$f_{\lambda}$ [$10^{-19}$ erg s$^{-1}$ cm$^{-2}$]', fontsize=12)
        fig.get_axes()[0].set_xlabel(r'$\lambda_{\mathrm{obs}}$', fontsize=12)
        fig.get_axes()[1].set_xlabel(r'$z$', fontsize=12)
        fig.get_axes()[1].set_ylabel(r'$p(z)$', fontsize=12)
        fig.get_axes()[0].legend(handles=[Line2D([0], [0], ls='-', color='red', label=fr'$z$={z:.2f}', markerfacecolor='red'),
                                            Line2D([0], [0], marker=None, color='w', label=fr'ID={ID}'),
                                            Line2D([0], [0], marker=None, color='w', label=fr"$\chi^2$={reduced_chi2:.2f}")], loc='upper left', fontsize=12)
        pos0 = fig.get_axes()[0].get_position()
        pos1 = fig.get_axes()[1].get_position()

        non_empty_images = []
        non_empty_names = []
        for image, name in zip(images_data, f_names):
            cutout = Cutout2D(image, SkyCoord(ra=RA*u.degree, dec=DEC*u.degree), u.Quantity((4, 4), u.arcsec), wcs=WCS(hdul[0].header))
            if not np.all(cutout.data == 0):
                non_empty_images.append(cutout)
                non_empty_names.append(name)
        
        if 'F444W' in non_empty_names:
            match_template_arr.append(non_empty_images[-1])
        
        n_images = len(non_empty_images)
        if max(nusefilt) > 18:
            n_cols = 7
        else:
            n_cols = 6
        n_rows = (n_images + n_cols - 1) // n_cols

        gs = GridSpec(n_rows, n_cols, figure=fig, wspace=0)
        for i, (cutout, name) in enumerate(zip(non_empty_images, non_empty_names)):
            ax = fig.add_subplot(gs[i // n_cols, i % n_cols])
            ax.imshow(cutout.data, origin='lower', cmap='hot', vmin=0, vmax=3*np.std(cutout.data))
            ax.set_title(name, fontsize=12, y=0.95)
            ax.axis('off')
            center_x, center_y = cutout.shape[1] / 2, cutout.shape[0] / 2
            length = min(cutout.shape) / 6
            gap = length * 0.6
            ax.plot([center_x - length, center_x - gap], [center_y, center_y], color='white', lw=1)  # Left horizontal line
            ax.plot([center_x + gap, center_x + length], [center_y, center_y], color='white', lw=1)  # Right horizontal line
            ax.plot([center_x, center_x], [center_y - length, center_y - gap], color='white', lw=1)  # Bottom vertical line
            ax.plot([center_x, center_x], [center_y + gap, center_y + length], color='white', lw=1)  # Top vertical line
        
        fig.subplots_adjust(bottom=pos0.y1 + 0.05, top=pos0.y1 + 0.9, left=pos0.x0 - 0.1, right=pos1.x1)
        plt.savefig(fig_path / f'eazy_fits/gds_photoz_fit_{str(ID)}.pdf', dpi=450, bbox_inches='tight')
    
```

```python
import cv2

def preprocess_image(image_data):
    # Normalize the image data to the range [0, 1]
    image_data = image_data / np.max(image_data)
    # Convert to 8-bit image (required by OpenCV)
    image_data = (image_data * 255).astype(np.uint8)
    return image_data

def detect_diffraction_spikes(image_data, psf_template):
    # Perform template matching
    result = cv2.matchTemplate(image_data, psf_template, cv2.TM_CCOEFF_NORMED)
    # Get the maximum correlation value
    _, max_val, _, _ = cv2.minMaxLoc(result)
    return max_val

def classify_based_on_spikes(max_val, threshold=0.5):
    return 'star' if max_val > threshold else 'galaxy'

cols = 5
rows = (len(cat['ID']) + cols - 1) // cols
with plt.ion():
    fig, axes = plt.subplots(rows, cols, figsize=(cols*3, rows*3), dpi=450)
    axes = axes.flatten()

    classifications = []
    psf_template = preprocess_image(fits.open(f_path / 'f444w.fits')[0].data)
    for ax, (cutout, ID) in zip(axes, zip(match_template_arr, cat['ID'])):
        image_data = preprocess_image(cutout.data)
        max_val = detect_diffraction_spikes(image_data, psf_template)
        classification = classify_based_on_spikes(max_val)
        classifications.append(classification)
        # print(f"The image of ID {ID} contains a {classification}.")
        ax.imshow(cutout.data, origin='lower', cmap='hot', vmin=0, vmax=3*np.std(cutout.data))
        ax.axis('off')
        ax.text(0.05, 0.95, f"ID: {ID}\n{classification}", color='white', 
            fontsize=14, weight='bold', ha='left', va='top', transform=ax.transAxes)
    
    print(f"Percentage of objects classified as stars: {classifications.count('star') / len(classifications) * 100:.2f}%")

    for i in range(len(cat['ID']), len(axes)):
        fig.delaxes(axes[i])
    plt.tight_layout()
    plt.savefig(fig_path / 'eazy_fits/gds_photoz_SG_class_v2.pdf', dpi=450, bbox_inches='tight')
    plt.show()
```
